```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


data = {
    'Height': [150,160,170,180,190],
    'Weight': [50,60,65,75,85],
    'Age': [20,25,30,35,40]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)


pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("\nReduced Data")
print(X_pca)

print("\nExplained Variance")
print(pca.explained_variance_ratio_)

print("\nEigenvalues")
print(pca.explained_variance_)


print("\nEigenvectors")
print(pca.components_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'Hours': [1, 2, 3, 4, 5],
    'Marks': [35, 45, 50, 60, 70]
}

df = pd.DataFrame(data)

print("Dataset")
print(df)


x = df[['Hours']]


y = df['Marks']


model = LinearRegression()


model.fit(x, y)
new_data = pd.DataFrame({
    'Hours': [6]
})

prediction = model.predict(new_data)



print("\nPredicted Marks for 6 Hours:", prediction[0])


print("\nSlope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression


data = {
    'Hours': [1, 2, 3, 4, 5],
    'Attendance': [60, 70, 80, 90, 95],
    'Previous_Marks': [40, 45, 55, 60, 70],
    'Final_Marks': [45, 52, 60, 68, 78]
}

df = pd.DataFrame(data)

print("Dataset")
print(df)


X = df[['Hours', 'Attendance', 'Previous_Marks']]


y = df['Final_Marks']


model = LinearRegression()


model.fit(X, y)


new_student = pd.DataFrame({
    'Hours': [6],
    'Attendance': [96],
    'Previous_Marks': [75]
})

prediction = model.predict(new_student)

print("\nPredicted Final Marks:", prediction[0])

print("\nCoefficients:")
for feature, coef in zip(X.columns, model.coef_):
    print(f"{feature}: {coef}")

print("\nIntercept:", model.intercept_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures


data = {
    'Years': [1, 2, 3, 4, 5],
    'Salary': [30000, 35000, 50000, 80000, 120000]
}

df = pd.DataFrame(data)

print("Dataset")
print(df)


X = df[['Years']]


y = df['Salary']

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)


model = LinearRegression()


model.fit(X_poly, y)


new_data = pd.DataFrame({
    'Years': [6]
})

new_data_poly = poly.transform(new_data)

prediction = model.predict(new_data_poly)

print("\nPredicted Salary:", prediction[0])

print("\nCoefficients:", model.coef_)
print("Intercept:", model.intercept_)
```
