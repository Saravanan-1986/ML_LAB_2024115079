```python
import pandas as pd
def find_s(examples,target):
    hypothesis=['0']*len(examples[0])
    for i in range (len(examples)):
        if target[i]=="yes":
            for j in range(len(hypothesis)):
                if hypothesis[j]=='0':
                    hypothesis[j]=examples[i][j]
                elif hypothesis[j]!=examples[i][j]:
                    hypothesis[j]='?'
    return hypothesis
data = {
    "sky":["sunny","sunny","rainy","sunny"],
    "airTemp":["warm","warm","cold","warm"],
    "humidity":["normal","high","high","high"],
    "wind":["strong","strong","strong","strong"],
    "water":["warm","warm","warm","cool"],
    "forecast":["same","same","change","same"],
    "enjoySport":["yes","yes","no","yes"]
}

df=pd.DataFrame(data)
examples=df.iloc[:, :-1].values.tolist()
target=df.iloc[:, -1].values.tolist()
hypothesis=find_s(examples,target)
print("find hypothesis")
print(hypothesis)







```

```python
import pandas as pd

def candidate_elimination(examples, target):
    num_attributes = len(examples[0])

   
    S = ['0'] * num_attributes
    G = [['?'] * num_attributes]

    for i in range(len(examples)):
        if target[i] == "yes":
           
            G = [g for g in G if all(g[j] == '?' or g[j] == examples[i][j]
                                     for j in range(num_attributes))]

            
            for j in range(num_attributes):
                if S[j] == '0':
                    S[j] = examples[i][j]
                elif S[j] != examples[i][j]:
                    S[j] = '?'

        else:  
            new_G = []

            for g in G:
                
                if all(g[j] == '?' or g[j] == examples[i][j]
                       for j in range(num_attributes)):

                    for j in range(num_attributes):
                        if S[j] != '?' and S[j] != '0' and S[j] != examples[i][j]:
                            new_hypothesis = g.copy()
                            new_hypothesis[j] = S[j]
                            if new_hypothesis not in new_G:
                                new_G.append(new_hypothesis)
                else:
                    new_G.append(g)

            G = new_G

        print(f"\nAfter Example {i+1}")
        print("S =", S)
        print("G =")
        for g in G:
            print(g)

    return S, G



data = {
    "sky": ["sunny", "sunny", "rainy", "sunny"],
    "airTemp": ["warm", "warm", "cold", "warm"],
    "humidity": ["normal", "high", "high", "high"],
    "wind": ["strong", "strong", "strong", "strong"],
    "water": ["warm", "warm", "warm", "cool"],
    "forecast": ["same", "same", "change", "same"],
    "enjoySport": ["yes", "yes", "no", "yes"]
}

df = pd.DataFrame(data)

examples = df.iloc[:, :-1].values.tolist()
target = df.iloc[:, -1].values.tolist()

S, G = candidate_elimination(examples, target)

print("\nFinal Specific Hypothesis (S):")
print(S)

print("\nFinal General Hypotheses (G):")
for g in G:
    print(g)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression


data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 50, 65, 80]
}

df = pd.DataFrame(data)

X = df[["Hours"]]
y = df["Marks"]


model = LinearRegression()
model.fit(X, y)


pred = model.predict([[6]])

print("Predicted Marks:", pred[0])
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Experience": [1, 2, 3, 4, 5],
    "Age": [22, 24, 26, 28, 30],
    "Score": [60, 70, 75, 85, 90],
    "Salary": [25000, 35000, 45000, 55000, 65000]
}

df = pd.DataFrame(data)

X = df[["Experience", "Age", "Score"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6, 32, 95]])

print("Predicted Salary:", prediction[0])
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

data = {
    "Years": [1, 2, 3, 4, 5],
    "Salary": [25000, 30000, 40000, 55000, 80000]
}

df = pd.DataFrame(data)

X = df[["Years"]]
y = df["Salary"]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)


model = LinearRegression()
model.fit(X_poly, y)

new_data = poly.transform([[6]])
prediction = model.predict(new_data)

print("Predicted Salary:", prediction[0])
```

```python
import pandas as pd
from sklearn.linear_model import Ridge


data = {
    "Experience": [1, 2, 3, 4, 5],
    "Age": [22, 24, 26, 28, 30],
    "Salary": [25000, 35000, 45000, 55000, 65000]
}

df = pd.DataFrame(data)

X = df[["Experience", "Age"]]
y = df["Salary"]


model = Ridge(alpha=1.0)

model.fit(X, y)

prediction = model.predict([[6, 32]])

print("Predicted Salary:", prediction[0])
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```
